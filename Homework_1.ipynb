```python
# Mounting Google Drive and reading CSV file
drive.mount('/content/drive')
df = pd.read_csv('/content/drive/MyDrive/ECGR_4105/Datasets/D3.csv')

# Initializing variables
X1 = df.values[:, 0]
X2 = df.values[:, 1]
X3 = df.values[:, 2]
Y = df.values[:, 3]
```

```python
### Problem 1 ###

iterations = 1000
learning_rate = 0.1
m = len(Y)

# Linear Regression (Gradient Descent) Algorithm

def linear_regression(X):
  theta_0 = 0.0
  theta_1 = 0.0
  cost_history = []

  for i in range(iterations):
    H = theta_0 + (theta_1 * X)

    cost = (1 / (2 * m)) * np.sum((H - Y) ** 2)
    cost_history.append(cost)

    d_theta_0 = (1 / m) * np.sum(H - Y)
    d_theta_1 = (1 / m) * np.sum((H - Y) * X)

    theta_0 = theta_0 - (learning_rate * d_theta_0)
    theta_1 = theta_1 - (learning_rate * d_theta_1)

  return theta_0, theta_1, cost_history

# Linear Models

theta_0_1, theta_1_1, cost_history_1 = linear_regression(X1)
print("Linear Model for X1: Y = ", theta_0_1, " + (", theta_1_1, ") X")

theta_0_2, theta_1_2, cost_history_2 = linear_regression(X2)
print("Linear Model for X2: Y = ", theta_0_2, " + (", theta_1_2, ") X")

theta_0_3, theta_1_3, cost_history_3 = linear_regression(X3)
print("Linear Model for X3: Y = ", theta_0_3, " + (", theta_1_3, ") X")

# Plotting Regression and Loss Curves

plt.subplot(2, 3, 1)
plt.scatter(X1, Y, color = "red", marker = "+")
plt.plot(X1, theta_0_1 + (theta_1_1 * X1))
plt.title("X1 Regression")

plt.subplot(2, 3, 2)
plt.scatter(X2, Y, color = "red", marker = "+")
plt.plot(X2, theta_0_2 + (theta_1_2 * X2))
plt.title("X2 Regression")

plt.subplot(2, 3, 3)
plt.scatter(X3, Y, color = "red", marker = "+")
plt.plot(X3, theta_0_3 + (theta_1_3 * X3))
plt.title("X3 Regression")

plt.subplot(2, 3, 4)
plt.plot(cost_history_1)
plt.title("X1 Cost")

plt.subplot(2, 3, 5)
plt.plot(cost_history_2)
plt.title("X2 Cost")

plt.subplot(2, 3, 6)
plt.plot(cost_history_3)
plt.title("X3 Cost")

# Final Cost Comparison

print("Final Cost for X1: ", cost_history_1[-1])
print("Final Cost for X2: ", cost_history_2[-1])
print("Final Cost for X3: ", cost_history_3[-1])
```

```python
### Problem 2 ###

iterations = 1000
learning_rate = 0.1
m = len(Y)

# Multiple Linear Regression (Gradient Descent) Algorithm

def multiple_linear_regression():
  theta_0 = 0.0
  theta_1 = 0.0
  theta_2 = 0.0
  theta_3 = 0.0
  cost_history = []

  for i in range(iterations):
    H = theta_0 + (theta_1 * X1) + (theta_2 * X2) + (theta_3 * X3)

    cost = (1 / (2 * m)) * np.sum((H - Y) ** 2)
    cost_history.append(cost)

    d_theta_0 = (1 / m) * np.sum(H - Y)
    d_theta_1 = (1 / m) * np.sum((H - Y) * X1)
    d_theta_2 = (1 / m) * np.sum((H - Y) * X2)
    d_theta_3 = (1 / m) * np.sum((H - Y) * X3)

    theta_0 = theta_0 - (learning_rate * d_theta_0)
    theta_1 = theta_1 - (learning_rate * d_theta_1)
    theta_2 = theta_2 - (learning_rate * d_theta_2)
    theta_3 = theta_3 - (learning_rate * d_theta_3)

  return theta_0, theta_1, theta_2, theta_3, cost_history

# Linear Model

theta_0_4, theta_1_4, theta_2_4, theta_3_4, cost_history_4 = multiple_linear_regression()
print("Linear Model for X1: Y = ", theta_0_4, " + (", theta_1_4, ") X1 + (", theta_2_4, ") X2 + (", theta_3_4, ") X3")

# Plotting Regression and Loss Curves

plt.subplot(2, 1, 1)
scatter_H = theta_0_4 + (theta_1_4 * X1) + (theta_2_4 * X2) + (theta_3_4 * X3)
plt.scatter(scatter_H, Y, color = "red", marker = "+")
plt.plot(scatter_H, theta_0_4 + (theta_1_4 * X1) + (theta_2_4 * X2) + (theta_3_4 * X3))
plt.title("X1, X2, X3 Regression")

plt.subplot(2, 1, 2)
plt.plot(cost_history_4)
plt.title("X1, X2, X3 Cost")

# Final Cost

print("Final Cost for X1, X2, X3: ", cost_history_4[-1])
```
